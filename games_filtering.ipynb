```python
import pandas as pd

df = pd.read_csv('steamspy_games.csv')
```

```python
# Remove the records in the dataframe that have empty values

df = df.dropna()

print(df)

df.to_csv('steamspy_games_filtered.csv', index=False)
```

```python
# Store the inappropriate words from the file "inappropriate_words.txt" into a list

f = open("inappropriate_words.txt", "r")

inappropriate_words = []

while 1:
    x = f.readline()
    if x == "":
        break
    x = x.replace("\n", "")
    inappropriate_words.append(x)

print("Words stored successfully in the list")
```

```python
for game in df['name']:
    for val in inappropriate_words:
        if val.lower() in game.lower():
            print(game)
```
