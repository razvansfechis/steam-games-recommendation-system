```python
import pandas as pd
import requests
import time
```

```python
# Initialize the tags column

for entry in df:
    df['tags'] = None

print(df)

df.to_csv('../data/3_steamspy_games_with_tags.csv', index=False)
```

```python
# The function fetches the detail of games from the given URL

def fetch_data(url, params=None):
    """
        Create a function to fetch the data from the steamspi's API

    :param url: the url of the steamspy API
    :param params: the parameters to be used in the request
    :return:
    """

    response = requests.get(url, params=params)

    steamspy_api_data = response.json()

    return steamspy_api_data
```

```python
def add_games_tags_to_df(url, params=None, current_appid=None):
    data = fetch_data(url, params)

    tags_list = []

    for tag in data['tags']:
        tags_list.append(tag)

    tags_list_as_string = ','.join(tags_list)

    df.loc[df['appid'] == int(current_appid), 'tags'] = tags_list_as_string

    return None
```

```python
# Fetch the tags for each game from SteamSPI's APIs

url = "https://steamspy.com/api.php" # The URL from where the games will be fetched

parameters = {"request": "appdetails", "appid": "730"} # The parameters that will be added to the URL

df = pd.read_csv('../data/3_steamspy_games_with_tags.csv')

count = 0

# 6100

# 25047

start_index = 82550

for i in range(start_index, len(df)):
    appid = df['appid'].iloc[i]

    print(f"Processing game number {i} with the appid: {appid}")

    parameters.update({'appid': str(appid)})
    count += 1

    add_games_tags_to_df(url, parameters, appid)
    time.sleep(0.1)

    df.to_csv('../data/3_steamspy_games_with_tags.csv', index=False)
```
