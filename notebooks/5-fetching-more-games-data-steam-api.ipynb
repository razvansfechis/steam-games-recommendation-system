```python
# The function fetches the detail of games from the given URL

def fetch_data(url, params=None):
    """
        Create a function to fetch the data from the steamspi's API

    :param url: the url of the steamspy API
    :param params: the parameters to be used in the request
    :return:
    """

    response = requests.get(url, params=params)

    steamspy_api_data = response.json()

    return steamspy_api_data
```

```python
def add_more_games_data(url, params=None, current_appid=None):
    data = fetch_data(url, params)

    df['no_owners'] = df['no_owners'].astype(str)

    try:
        df.loc[df['appid'] == int(current_appid), 'no_owners'] = data["owners"]
        df.loc[df['appid'] == int(current_appid), 'median_playtime'] = data["median_forever"]
        df.loc[df['appid'] == int(current_appid), 'no_positive_ratings'] = data["positive"]
        df.loc[df['appid'] == int(current_appid), 'no_negative_ratings'] = data["negative"]

    except TypeError as te:
        print(te)

    return None
```

```python
# Fetch the tags for each game from SteamSPI's APIs

url = "https://steamspy.com/api.php" # The URL from where the games will be fetched

parameters = {"request": "appdetails", "appid": "730"} # The parameters that will be added to the URL

df = pd.read_csv('../data/6_steam_games_more_data.csv')

count = 0

start_index = 81890 #81931

for i in range(start_index, len(df)):
    appid = df['appid'].iloc[i]

    print(f"Processing game number {i} with the appid: {appid}")

    parameters.update({'appid': str(appid)})
    count += 1

    add_more_games_data(url, parameters, appid)

    #if count == 100:
    df.to_csv('../data/6_steam_games_more_data.csv', index=False)
    # count = 0
```

```python
# Initializing a new column called avg_no_owners, where will be the average number of owners for each game

for entry in df:
    df['avg_no_owners'] = 0

df['avg_no_owners'] = df['avg_no_owners'].astype(int)

print(df)
```

```python
def get_average_no_owners(no_owners_as_string):
    """
    Use this function to get the average number of owner for a game

    :param no_owners_as_string:
    :return:
    """

    for word in no_owners_as_string:
        if word == "," or word == ".":
            no_owners_as_string = no_owners_as_string.replace(word, '')

    lower_interval_value = 0
    higher_interval_value = 0
    iterator = 0

    while no_owners_as_string[iterator] != ' ':
        lower_interval_value = lower_interval_value * 10 + int(no_owners_as_string[iterator])
        iterator += 1

    while no_owners_as_string[iterator] == ' ':
        iterator += 1

    while iterator < len(no_owners_as_string):
        higher_interval_value = higher_interval_value * 10 + int(no_owners_as_string[iterator])
        iterator += 1

    average_number_of_owners = int((lower_interval_value + higher_interval_value) / 2)

    return average_number_of_owners

```

```python
# Adding the average number of owners to the .csv

df = pd.read_csv('../data/6_steam_games_more_data.csv')

for i in range(0, len(df)):
    no_owners = df['no_owners'].iloc[i]
    current_appid = df['appid'].iloc[i]

    avg_no_owners = get_average_no_owners(no_owners)

    print("Processing game number " + str(i) + " with the appid: " + str(current_appid))

    df.loc[df['appid'] == int(current_appid), 'avg_no_owners'] = avg_no_owners

    if i == len(df) - 1:
        df.to_csv('../data/6_steam_games_more_data.csv', index=False)
```

```python
# Initializing a new column called likes_percentile, where will be the average number of owners for each game

for entry in df:
    df['positive_ratings_percentile'] = 0

df['positive_ratings_percentile'] = df['positive_ratings_percentile'].astype(int)

print(df)
```

```python
def get_positive_ratings_percentile(no_positive_ratings, no_negative_ratings):
    positive_ratings_percentile = no_positive_ratings / (no_positive_ratings + no_negative_ratings) * 100

    return int(positive_ratings_percentile)
```

```python
# Adding the percentile of positive reviews to the .csv

df = pd.read_csv('../data/6_steam_games_more_data.csv')

for i in range(0, len(df)):
    current_appid = df['appid'].iloc[i]

    no_positive_ratings = df['no_positive_ratings'].iloc[i]
    no_negative_ratings = df['no_negative_ratings'].iloc[i]

    if no_positive_ratings == 0:
        positive_ratings_percentile = 0
    else:
        positive_ratings_percentile = get_positive_ratings_percentile(no_positive_ratings, no_negative_ratings)

    print("Processing game number " + str(i) + " with the appid: " + str(current_appid))

    df.loc[df['appid'] == int(current_appid), 'positive_ratings_percentile'] = positive_ratings_percentile

    if i == len(df) - 1:
       df.to_csv('../data/6_steam_games_more_data.csv', index=False)
```
