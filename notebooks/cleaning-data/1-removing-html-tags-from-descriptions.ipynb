```python
import pandas as pd
from bs4 import BeautifulSoup
import concurrent.futures
```

```python
df = pd.read_csv('../../data/6_steam_games_more_data.csv')

counter = 0
```

```python
def get_cleaned_detailed_description(detailed_description):
    """
    Get cleaned detailed description from a BeautifulSoup object (BeautifulSoup with 'html.parser' removes all the HTML tags)

    :param detailed_description: the description that needs to be cleaned
    :return: The cleaned description as a string
    """
    global counter
    counter += 1

    print(f"Cleaning game number: {counter}")

    soup = BeautifulSoup(detailed_description, 'html.parser')

    return soup.get_text()
```

```python
# Remove the HTML tags from all the detailed descriptions using threads for faster processing

detailed_descriptions = []
cleaned_detailed_descriptions = []

for i in range(len(df)):
    detailed_descriptions.append(df.iloc[i]['detailed_description'])

with concurrent.futures.ThreadPoolExecutor(max_workers=8) as executor:
    cleaned_detailed_descriptions = executor.map(get_cleaned_detailed_description, detailed_descriptions)

df['detailed_description'] = list(cleaned_detailed_descriptions)

df.to_csv('../../data/7_steam_games_cleaned_data.csv', index=False)
```
