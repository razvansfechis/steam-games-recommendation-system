```python
import pandas as pd
```

```python
# Adding the value "." to the column detailed_description to games that have no detailed description

df = pd.read_csv('../../data/7_steam_games_cleaned_data.csv')

df['detailed_description'] = df['detailed_description'].fillna('.') # use a "." that is a special character which will be removed
                                                                    # in the column that will be used by the ML

print(df[df['detailed_description'] == '.'])

df.to_csv('../../data/7_steam_games_cleaned_data.csv', index=False)
```

```python
# Create a new column called combined_cleaned_descriptions where there is the combined values of the detailed_description and the short_description, and it will be cleaned and ready to use by the ML

df = pd.read_csv('../../data/7_steam_games_cleaned_data.csv')

df['combined_description'] = df['detailed_description'] + " " + df['short_description']

print(df)

df.to_csv('../../data/7_steam_games_cleaned_data.csv', index=False)
```
