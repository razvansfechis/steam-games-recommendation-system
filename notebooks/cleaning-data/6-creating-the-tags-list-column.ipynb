```python
import pandas as pd
```

```python
df = pd.read_csv("../../data/12_steam_games_lemmatized_descriptions.csv")
```

```python
def modify_tags(tag_string):
    modified_tags = []

    for tag in tag_string.split(','):
        tag = tag.strip()
        if tag:
            tag = tag.lower().replace(" ", "-") # replace " " with - and convert to lowercase
            # ex. Online Multi-Player -> online-multi-player
            modified_tags.append(tag)

    return modified_tags

df['tag_list'] = df['tags'].apply(modify_tags)

print(df['tag_list'].head())

df.to_csv("../../data/13_steam_games_tags_list.csv", index=False)
```
