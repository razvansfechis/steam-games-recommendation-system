```python
import pandas as pd
```

```python
df = pd.read_csv("../../data/12_steam_games_lemmatized_descriptions.csv")
```

```python
number_of_games_before_removing_duplicates = len(df)

print("Number of games before removing duplicates: ", number_of_games_before_removing_duplicates)

df.drop_duplicates(subset=['name'], keep='first', inplace=True)

number_of_games_after_removing_duplicates = len(df)

print("Number of games after removing duplicates: ", number_of_games_after_removing_duplicates)

df.to_csv("../../data/13_steam_games_final.csv", index=False)
```
