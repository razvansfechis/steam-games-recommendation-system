```python
import pandas as pd
import langid
from deep_translator import GoogleTranslator
import time
```

```python
counter = 0
```

```python
def get_translated_short_description(appid, df):
    """
    Translates the description for games in chunks of 200 characters chunks

    return: the translated game description
    """

    global counter
    counter += 1

    print(f"Translating game no. {counter} with appid: {appid}")

    description = df.loc[df['appid'] == appid, 'translated_short_description'].iloc[0]

    description_chunks = []

    for i in range(0, len(description), 200):
        description_chunks.append(description[i:i + 200])

    translated_chunks = []
    for chunk in description_chunks:
        try:
            translated_chunk = GoogleTranslator(source='auto', target='en').translate(chunk)
        except Exception as e:
            print(f"Error {e} for the appid: {appid}")
            translated_chunk = ""

        if translated_chunk is None:
            translated_chunk = ""  # or fallback to original chunk

        translated_chunks.append(translated_chunk)

        time.sleep(1)

    translated_description = ""

    translated_description += (str(appid) + " ")

    for translated_chunk in translated_chunks:
        translated_description += translated_chunk

    return translated_description
```

```python
def get_foreign_short_descriptions_appids(df):
    """
    Get all the appids for games with foreign language in the description
    :return: all the appids for games with foreign language in the description
    """
    foreign_short_descriptions_appids = []

    for i in range(len(df)):
        description = df.iloc[i]['translated_short_description']

        if pd.isna(description):
            continue

        lang, _ = langid.classify(df.iloc[i]['translated_short_description'])

        if lang != 'en':
            foreign_short_descriptions_appids.append(int(df.iloc[i]['appid']))

    print("Appids for games with foreign description: ", foreign_short_descriptions_appids)
    return foreign_short_descriptions_appids
```

```python
# Translating the short_description and saving the translated data in a new csv

df = pd.read_csv('../../data/13_steam_games_final.csv')

df['translated_short_description'] = df['short_description']

translated_short_descriptions = []

foreign_short_descriptions_appids = get_foreign_short_descriptions_appids(df)

for appid in foreign_short_descriptions_appids:
    translated_description = get_translated_short_description(appid, df)
    translated_short_descriptions.append(translated_description)

for translated_short_description in translated_short_descriptions:
    appid = translated_short_description.split()[0]
    translated_description = translated_short_description.split(" ", 1)[1]

    df.loc[df['appid'] == int(appid), 'translated_short_description'] = translated_description

df.to_csv('../../data/14_steam_games_final.csv', index=False)
```
