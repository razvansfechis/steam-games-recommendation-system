```python
import time
import requests
import pandas as pd
```

```python
# The function that fetches the games from the given URL

def fetch_data(url, params=None):
    """
        Create a function to fetch the data from the steamspi's API

    :param url: the url of the steamspy API
    :param params: the parameters to be used in the request
    :return:
    """

    response = requests.get(url, params=params)

    steamspy_api_data = response.json()

    return steamspy_api_data
```

```python
url = "https://steamspy.com/api.php" # The URL from where the games will be fetched

parameters = {"request": "all", "page": "0"} # The parameters that will be added to the URL

print("Getting the data from the nr. 0 page")

all_steamspy_api_data = fetch_data(url, parameters)
time.sleep(61) # The allowed pool rate (the frequency of requests) is 1 request per 60 seconds for "all" requests

# Getting the games from 87 pages (pages from number 0 till number 86)

for i in range(1, 87):
    print(f"Getting the data from the nr. {i} page")
    parameters.update({"page": str(i)})

    print("PARAMETERS", parameters)

    all_steamspy_api_data.update(fetch_data(url, parameters))
    time.sleep(61)

# print(all_steamspy_api_data)
# print(len(all_steamspy_api_data))
```

```python
df = pd.DataFrame.from_dict(all_steamspy_api_data, orient='index')

df = df[['appid', 'name']]

print(df)

df.to_csv("1_steamspy_games.csv", index=False)
```
