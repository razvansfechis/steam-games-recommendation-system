```python
import pandas as pd
```

```python
# Get the DataFrame from the .csv file

df = pd.read_csv('../../data/1_steamspy_games.csv')
```

```python
# Remove the records in the dataframe that have empty values

df = df.dropna()

print(df)

df.to_csv('2_steamspy_games_filtered.csv', index=False)
```

```python
# Store the inappropriate words from the file "inappropriate_words.txt" into a list

f = open("../../data/inappropriate_words.txt", "r")

inappropriate_words = []

while 1:
    x = f.readline()
    if x == "":
        break
    x = x.replace("\n", "")
    inappropriate_words.append(x)

print("Words stored successfully in the list")
```

```python
# Remove the games from the DataFrame that have an inappropriate word in the name and store the new DataFrame into the csv file

for game_name in df['name']:
    for val in inappropriate_words:
        if val.lower() in game_name.lower():
            df.drop(df[df['name'] == game_name].index, inplace=True)

print(df)

df.to_csv('2_steamspy_games_filtered.csv', index=False)
```
