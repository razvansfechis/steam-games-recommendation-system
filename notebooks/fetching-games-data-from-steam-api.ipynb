```python
import pandas as pd
import time
import requests
```

```python
df = pd.read_csv("../data/steamspy_games_with_tags_no_empty.csv")

print(df)
```

```python
# Initializing the detailed_description, short_description & header_image columns

for entry in df:
    df['detailed_description'] = None
    df['short_description'] = None
    df['header_image_url'] = None

print(df)
```

```python
# The function fetches the detail of games from the given URL

def fetch_data(url, params=None):
    """
        Create a function to fetch the data from the steamspi's API

    :param url: the url of the steamspy API
    :param params: the parameters to be used in the request
    :return:
    """

    response = requests.get(url, params=params)

    steam_api_data = response.json()

    return steam_api_data
```

```python
def add_games_data(url, params=None, current_appid=None):
    data = fetch_data(url, params)

    try:
        if data[str(appid)]['success'] and data[str(appid)] is not None:
            df.loc[df['appid'] == int(current_appid), 'name'] = data[str(appid)]["data"]["name"]
            df.loc[df['appid'] == int(current_appid), 'detailed_description'] = data[str(appid)]["data"]["detailed_description"]
            df.loc[df['appid'] == int(current_appid), 'short_description'] = data[str(appid)]["data"]["short_description"]
            df.loc[df['appid'] == int(current_appid), 'header_image_url'] = data[str(appid)]["data"]["header_image"]

    except TypeError:
        print("Error, moving on")

    return None
```

```python
url = "https://store.steampowered.com/api/appdetails" # The URL from where the games will be fetched

parameters = {"appids": "730"} # The parameters that will be added to the URL

count = 0

start_index = 1880

for i in range(start_index, len(df)):
    appid = df['appid'].iloc[i]

    print(f"Processing game number {i} with the appid: {appid}")

    parameters.update({'appids': str(appid)})
    count += 1

    time.sleep(1)

    add_games_data(url, parameters, appid)

    if count == 100:
        df.to_csv('../data/steam_games_data.csv', index=False)
        count = 0
```
